```python
import pandas as pd

data = {
    "Date": [
        "2025-05-01","2025-05-02","2025-05-03","2025-05-04",
        "2025-05-05","2025-05-06","2025-05-07","2025-05-08",
        "2025-05-09","2025-05-10"
    ],
    "Category": [
        "Food","Transport","Shopping","Food",
        "Bills","Entertainment","Transport","Food",
        "Shopping","Bills"
    ],
    "Amount": [250,100,1200,300,1500,500,150,400,800,1000]
}

df = pd.DataFrame(data)

df.to_csv("expenses.csv", index=False)

print("Dataset Created Successfully!")
df.head()
```

```python
import pandas as pd

df = pd.read_csv("expenses.csv")

print("Expense Data:")
print(df)

total_expense = df["Amount"].sum()

print("\nTotal Expense:", total_expense)

category_expense = df.groupby("Category")["Amount"].sum()

print("\nExpense By Category:")
print(category_expense)

highest = category_expense.idxmax()

print("\nHighest Spending Category:", highest)
```

```python
import matplotlib.pyplot as plt

category_expense = df.groupby("Category")["Amount"].sum()

plt.figure(figsize=(8,6))
category_expense.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Expense Distribution")
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(8,5))

category_expense.plot(kind="bar")

plt.title("Expenses By Category")
plt.xlabel("Category")
plt.ylabel("Amount")

plt.show()
```
